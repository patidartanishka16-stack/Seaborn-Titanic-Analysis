### Project: Titanic — Passenger & Survival Analysis

#### Titanic Data Visualization using Seaborn

## 1. Introduction

This project explores the Titanic dataset using Python, Pandas, and Seaborn.
The analysis focuses on passenger class, survival, fare distribution, and
differences between passenger groups.

## 2. Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

## 3. Load Dataset

```python
titanic= pd.read_csv("Titanic-Dataset.csv")
```

## 4. Visualization and Analysis

Task 1️⃣ — Passenger Class Distribution

```python
import seaborn as sns

sns.countplot(x="Pclass",data=titanic)
plt.title("Passenger Class Distribution")
plt.show()
```

In Titanic dataset, The number of passengers in Pclass 3 is higher than in Pclass 1.

Task 2️⃣ — Survival by Sex

Question:

Graph se 1 finding likho.

```python
import seaborn as sns

sns.countplot(x="Sex",hue="Survived",data=titanic)
plt.title("Survival by Sex")
plt.show()
```

The number of non-survived passengers is higher than the number of survived passengers. Among females, the number of survivors is higher than among males.

Task 3️⃣ — Fare Distribution

Question:

Titanic passengers ke Fare ka distribution kaisa tha?

```python
import seaborn as sns

sns.histplot(x="Fare",data=titanic)
plt.title("Fare Distribution")
plt.show()
```

Most passenger fares are concentrated in the 0–100 range, while a smaller number of passengers have much higher fares.

Task 4️⃣ — Fare by Passenger Class

Kya different Pclass ke passengers ka Fare spread different hai?

```python
import seaborn as sns

sns.boxplot(x="Pclass",y="Fare",data=titanic)
plt.title("Fare by Passenger Class")
plt.show()
```

Pclass 1 has a higher median fare and a wider fare spread compared with Pclass 2 and Pclass 3.

```python
import seaborn as sns

sns.barplot(x="Pclass",y="Survived",data=titanic)
plt.title("Passenger vs Survival Distribution")
plt.show()
```

The survival rate is higher in the Pclass 1.

Pclass 1 has the highest survival rate among the three passenger classes.

The Titanic dataset shows differences in passenger distribution, survival, and fare across different groups. Pclass 3 had the largest number of passengers. Female passengers had a higher number of survivors compared with male passengers. Most fares were concentrated in the lower range, while Pclass 1 showed a higher fare distribution. Pclass 1 also had the highest survival rate.

## 5. Key Findings

- Pclass 3 had the highest number of passengers.
- Female passengers had a higher number of survivors compared with male passengers.
- Most passenger fares were concentrated in the lower fare range.
- Pclass 1 showed a higher fare distribution compared with other classes.
- Pclass 1 had the highest survival rate among the three passenger classes.
